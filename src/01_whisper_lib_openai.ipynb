# Usando lib Whisper
### Método mais simples

```python
# lib criada pela propria openai para realizar inferencia simples para whisper, com ela conseguimos ter acesso as metricas do modelos, e um fluxo mais otimizado.
import whisper
import torch

# porque usamos essas variaveis, e quais seus impactos ?
device = "cuda" if torch.cuda.is_available() else "cpu"

# carregamento do modelos para memoria.
model_orig = whisper.load_model("large-v3-turbo", device=device)

# executando inferencia, do modelos.
result = model_orig.transcribe(
    r"C:\Users\lucasmg-cogna\Dev\AI-Audio-Transcription-Examples\audio\simulacao-de-dialogo.wav",
    language="pt",
    verbose=False,
    fp16=True
)

# exibindo metricas e resultado
for seg in result["segments"]:
    confidence = "alta"   if seg["avg_logprob"]        > -0.3 else "baixa"
    no_speech  = "clara"  if seg["no_speech_prob"]     < 0.3  else "silêncio/ruído"
    halluc     = "ok"     if seg["compression_ratio"]  < 2.4  else "possível alucinação"
    fallback   = "nenhum" if seg["temperature"] == 0.0       else "ativado"

    no_speech_pct = seg["no_speech_prob"] * 100
    fala_pct      = 100 - no_speech_pct
    bar_len       = 20
    filled        = round(fala_pct / 100 * bar_len)

    print(f"[{seg['start']:.2f}s → {seg['end']:.2f}s]")
    print(f"  texto:               {seg['text'].strip()}")
    print(f"  transcrição:         {confidence} ({seg['avg_logprob']:.4f})")
    print(f"  presença de fala:    {no_speech} {fala_pct:.1f}%")
    print(f"  alucinação:          {halluc} (ratio: {seg['compression_ratio']:.4f})")
    print(f"  fallback:            {fallback} (temp: {seg['temperature']})")
    print()
```
